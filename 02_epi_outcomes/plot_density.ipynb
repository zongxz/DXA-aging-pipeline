```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from matplotlib.ticker import FuncFormatter
from matplotlib.colors import LinearSegmentedColormap

# 设置全局字体
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 16
plt.rcParams['axes.titlesize'] = 20
plt.rcParams['axes.labelsize'] = 22
plt.rcParams['legend.fontsize'] = 14

# 定义千分位格式化函数
def thousands_formatter(x, pos):
    return f'{int(x):,}'

# 由单色生成 colormap（白 -> 指定颜色）
def make_cmap(hex_color):
    return LinearSegmentedColormap.from_list('custom', ['#FFFFFF', hex_color])

# body part -> 颜色
part_colors = {
    'Knee': '#006affff',   # 蓝
    'Hip':       '#1dcc2fff',   # 绿
    'Knee - Hip':        '#FD3E5Cff',   # 红
}

# 性别颜色
male_color   = '#AD1AEE'
female_color = '#ff7e0f'

# 定义CSV文件列表（title 控制标题显示，并决定颜色）
csv_files = [
    {'file': 'double_right_oof.csv',              'name': 'double_right',       'label': 'Double Right',       'title': 'Knee - Hip'},
    {'file': 'single_right_knee_oof.csv',         'name': 'single_right_knee',  'label': 'Single Right Knee',  'title': 'Knee'},
    {'file': 'single_right_hip_oof.csv',          'name': 'single_right_hip',   'label': 'Single Right Hip',   'title': 'Hip'},
    {'file': 'double_left_external_valid.csv',    'name': 'double_left',        'label': 'Double Left',        'title': 'Knee - Hip'},
    {'file': 'single_left_knee_external_valid.csv','name': 'single_left_knee',  'label': 'Single Left Knee',   'title': 'Knee'},
    {'file': 'single_left_hip_external_valid.csv','name': 'single_left_hip',    'label': 'Single Left Hip',    'title': 'Hip'}
]

for csv_info in csv_files:
    file_path = f'../data/{csv_info["file"]}'
    base_name = csv_info['name']
    label = csv_info['label']
    plot_title = csv_info['title']

    # 根据 body part 取颜色
    base_color = part_colors[plot_title]
    custom_cmap = make_cmap(base_color)

    # 性别 colormap
    cmap_male   = make_cmap(male_color)
    cmap_female = make_cmap(female_color)

    plot_data = pd.read_csv(file_path)

    # 分离性别
    plot_data_male = plot_data[plot_data['sex'] == 1]
    plot_data_female = plot_data[plot_data['sex'] == 0]

    # ============ 图1：总图 ============
    fig, ax = plt.subplots(figsize=(6, 6))

    sns.kdeplot(
        x=plot_data.age,
        y=plot_data.adjusted_age,
        cmap=custom_cmap,
        fill=True,
        bw_adjust=.5,
        thresh=0.1,
        ax=ax
    )
    ax.plot(
        [min(plot_data['age']), max(plot_data['age'])],
        [min(plot_data['age']), max(plot_data['age'])],
        color='black', linestyle='--'
    )

    # 设置轴范围和刻度
    ax.set_xlim(38, 92)
    ax.set_ylim(38, 92)
    ax.set_xticks([40, 50, 60, 70, 80, 90])
    ax.set_yticks([40, 50, 60, 70, 80, 90])

    # 强制绘图区域为正方形
    ax.set_box_aspect(1)

    # 应用千分位格式化
    ax.xaxis.set_major_formatter(FuncFormatter(thousands_formatter))
    ax.yaxis.set_major_formatter(FuncFormatter(thousands_formatter))

    # 标题和标签（不显示样本量）
    ax.set_title(plot_title, fontweight='bold', fontsize=20)
    ax.set_xlabel('Chronological age (years)', fontweight='bold')
    ax.set_ylabel('Predicted age (years)', fontweight='bold')
    ax.tick_params(axis='both', which='major', width=2)

    plt.tight_layout()
    plt.savefig(f'../result/F1_密度图/{base_name}_total.png', dpi=600)
    plt.savefig(f'../result/F1_密度图/{base_name}_total.pdf')
    plt.close()

    # ============ 图2：性别分图（横向 1x2） ============
    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    # 男性
    if len(plot_data_male) > 0:
        sns.kdeplot(
            x=plot_data_male.age,
            y=plot_data_male.adjusted_age,
            cmap=cmap_male,
            fill=True,
            bw_adjust=.5,
            thresh=0.1,
            ax=axes[0]
        )
    axes[0].plot(
        [min(plot_data['age']), max(plot_data['age'])],
        [min(plot_data['age']), max(plot_data['age'])],
        color='black', linestyle='--'
    )
    axes[0].set_title('Male', fontweight='bold', fontsize=16)   # 小于 Knee 字号(20)
    axes[0].set_xlabel('Chronological age (years)', fontweight='bold')
    axes[0].set_ylabel('Predicted age (years)', fontweight='bold')
    axes[0].set_xlim(38, 92)
    axes[0].set_ylim(38, 92)
    axes[0].set_xticks([40, 50, 60, 70, 80, 90])
    axes[0].set_yticks([40, 50, 60, 70, 80, 90])
    axes[0].set_box_aspect(1)   # 正方形绘图区域
    axes[0].xaxis.set_major_formatter(FuncFormatter(thousands_formatter))
    axes[0].yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
    axes[0].tick_params(axis='both', which='major', width=2)

    # 女性
    if len(plot_data_female) > 0:
        sns.kdeplot(
            x=plot_data_female.age,
            y=plot_data_female.adjusted_age,
            cmap=cmap_female,
            fill=True,
            bw_adjust=.5,
            thresh=0.1,
            ax=axes[1]
        )
    axes[1].plot(
        [min(plot_data['age']), max(plot_data['age'])],
        [min(plot_data['age']), max(plot_data['age'])],
        color='black', linestyle='--'
    )
    axes[1].set_title('Female', fontweight='bold', fontsize=16)  # 小于 Knee 字号(20)
    axes[1].set_xlabel('Chronological age (years)', fontweight='bold')
    axes[1].set_ylabel('Predicted age (years)', fontweight='bold')
    axes[1].set_xlim(38, 92)
    axes[1].set_ylim(38, 92)
    axes[1].set_xticks([40, 50, 60, 70, 80, 90])
    axes[1].set_yticks([40, 50, 60, 70, 80, 90])
    axes[1].set_box_aspect(1)   # 正方形绘图区域
    axes[1].xaxis.set_major_formatter(FuncFormatter(thousands_formatter))
    axes[1].yaxis.set_major_formatter(FuncFormatter(thousands_formatter))
    axes[1].tick_params(axis='both', which='major', width=2)

    # 在两幅图顶部居中显示总标题
    fig.suptitle(plot_title, fontweight='bold', fontsize=20, y=0.95)

    plt.tight_layout(rect=[0, 0, 1, 0.94])
    plt.savefig(f'../result/F1_密度图/{base_name}_sex.png', dpi=600)
    plt.savefig(f'../result/F1_密度图/{base_name}_sex.pdf') 
    plt.close()

    print(f"完成 {label} (n = {len(plot_data):,})")
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

csv_files = [
    {'file': 'double_right_oof.csv',            'label': 'Double Right',       'title': 'Knee - Hip'},
    {'file': 'single_right_knee_oof.csv',       'label': 'Single Right Knee',   'title': 'Knee'},
    {'file': 'single_right_hip_oof.csv',        'label': 'Single Right Hip',    'title': 'Hip'},
    {'file': 'double_left_external_valid.csv',  'label': 'Double Left',         'title': 'Knee - Hip'},
    {'file': 'single_left_knee_external_valid.csv', 'label': 'Single Left Knee','title': 'Knee'},
    {'file': 'single_left_hip_external_valid.csv',  'label': 'Single Left Hip', 'title': 'Hip'},
]
DATA_DIR = '../data'

def eval_pair(y_true, y_pred):
    d = pd.DataFrame({'t': y_true, 'p': y_pred}).dropna()
    mse = mean_squared_error(d['t'], d['p'])
    mae = mean_absolute_error(d['t'], d['p'])
    r, _ = pearsonr(d['p'], d['t'])
    n = len(d)
    return n, mse, mae, r

# 每一列 = (Dataset, Model)
cols = {}
for cfg in csv_files:
    df = pd.read_csv(os.path.join(DATA_DIR, cfg['file']))
    for pred_col, tag in [('predicted_age', 'Unadjusted'),
                          ('adjusted_age',    'Adjusted')]:
        n, mse, mae, r = eval_pair(df['age'], df[pred_col])
        cols[(cfg['label'], tag)] = [n, round(mse,3), round(mae,3), round(r,3)]

result = pd.DataFrame(cols, index=['N','MSE','MAE','Pearson_r'])
result.columns = pd.MultiIndex.from_tuples(result.columns, names=['Dataset','Model'])
print(result.to_string())

result.to_csv('../result/F1_密度图/age_metrics_summary.csv')
```
